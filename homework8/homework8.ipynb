```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w*x)
```

```python
x = np.linspace(0, 2*np.pi, 1000)
```

```python
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
```

```python
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(25, 5))

for amp, freq in zip(amplitude, frequency):
    y1 = make_sine_wave(x, A=amp, w=1)
    axes[0].plot(x, y1, label=f"A = {amp}")

    y2 = make_sine_wave(x, A=1, w=freq)
    axes[1].plot(x, y2, label=f"w = {freq}")

    y3 = make_sine_wave(x, A=amp, w=freq)
    axes[2].plot(x, y3, label=f"A = {amp}, w = {freq}")

titles = [
    "Varying Amplitude (A)",
    "Varying Frequency (w)",
    "Varying Amplitude & Frequency",
]

for ax, title in zip(axes, titles):
    ax.set_title(title, fontsize=16)
    ax.set_xlabel("radians", fontsize=14)
    ax.set_ylabel("y = A · sin(wx)", fontsize=14)
    ax.legend(fontsize=11, loc="upper right")

plt.tight_layout()
plt.show()
```
