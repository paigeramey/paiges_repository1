```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/planets.csv"
planets = pd.read_csv(url)
```

```python
new_planets = planets.dropna(axis=1)
```

```python
new_planets.shape
```

```python
# 1036 rows, 4 collumns
```

```python
print(new_planets.head(5))
print(new_planets.tail(5))
```

```python
# first column tells the number planet
# second tells the method the planet was found with
# third shows the number of planets in that system
# fourth shows the year the planet was discovered
```

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(12, 10))
# plot 1: 
sns.histplot(
    data=new_planets, x="year", kde=True, ax=axes[0, 0], color="skyblue"
)
axes[0, 0].set_title("Distribution of Discovery Years")
# plot 2
sns.countplot(data=new_planets, x="number", ax=axes[0, 1], palette="magma")
axes[0, 1].set_title("Planets per Star System")

# plot 3 (Bottom-Left): Top discovery methods
top_methods = new_planets["method"].value_counts().head(5)
axes[1, 0].barh(top_methods.index, top_methods.values, color="teal")
axes[1, 0].set_title("Top 5 Discovery Methods")
axes[1, 0].set_xlabel("Count")

# plot 4 (Bottom-Right): Discoveries over time by top method
top_3_methods = new_planets["method"].value_counts().head(3).index
filtered_data = new_planets[new_planets["method"].isin(top_3_methods)]

sns.scatterplot(
    data=filtered_data,
    x="year",
    y="number",
    hue="method",
    ax=axes[1, 1],
    s=70,
)
axes[1, 1].set_title("Planets per System over Time (Top 3 Methods)")

# 3. Adjust spacing between plots and display
plt.tight_layout()
plt.show()
```
